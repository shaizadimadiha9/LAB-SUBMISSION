### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
  
# C1
def load_records(path):
    """Read label,value,limit lines from a file and return a list of dicts."""
    records = []
    try:
        #to open the file using with statement 
        with open(path) as f:
            #to read the file line by line
            for line in f:
                #to remove the unwanted whitespaces
                line = line.strip()
                #if the line is empty then skip and move to next line
                if not line:
                    continue
                label, value, limit = line.split(",")
                #to convert the numbers to floats and append dictionary to list  
                records.append({"label": label, 
                                "value": float(value),
                                  "limit": float(limit)})
    except FileNotFoundError:
        #handle the case where the file does not exist 
        print(f"{path} not found")
        # to return an empty list on failure 
        return []
    return records


# Test 1: load the sample file and print label and status, as in D16
for record in load_records("records_sample.txt"):
    if record["value"] > record["limit"]:
        print(record["label"] + ": OVER LIMIT")
    else:
        print(record["label"] + ": OK")

# Test 2: missing file
print(load_records("nope.txt"))
            
````
